Understand what you want to accomplish. This is the first step of the CRISP-DM framework for data science projects. Business understanding is about looking at the data in context. **a. Read the dataset description. Who produced tha data? When, how, and for what purpose?**

**Answer:** 

**Who:** a global peatland field survey led by peatland and soil-microbiology researchers at the University of Tartu Estonia with international partners. The greenhouse-gas and soil data are described in Pärn et al. (2018, Nature Communications) and Pärn et al. (2025, Scientific Reports); the soil metagenomes of the same stations are described in Bahram et al. (2022, Nature Communications). The table we use was compiled from this survey for the course and is not publicly released.

**When:** in several field campaigns. The first 58 sites were published in Pärn et al. (2018); the survey was later extended, and our table has 75 sites (as in Bahram et al. 2022 and Pärn et al. 2025). Each station was measured once (a snapshot); the exact dates per site are not part of our table.

**How:** 75 peatland sites on 6 continents, with 1–8 stations per site. At each station, gas fluxes were measured with closed chambers (CH4 in µg C m⁻² h⁻¹; negative values mean uptake), the water table, soil moisture and soil temperature were measured in the field, and soil from the chambers was pooled into one sample. The soil was analysed in the lab (pH, nutrients, C, N, peat decomposition), and its DNA was shotgun-sequenced. The DNA reads were assigned to microbial genera with the software Kaiju; the course table keeps 210 genera and two summary columns (share of methanogens and methanotrophs).

**Purpose:** to understand what controls the exchange of greenhouse gases (CH4, N2O, CO2) between peatlands and the atmosphere in natural and drained peatlands under different land uses, and how the soil microbiome is linked to it. In this project we use it to predict CH4 and to test whether microbial data add information beyond soil conditions.

**b. Name a plausible stakeholder, their goals, and the concrete decision the analysis would support. Who benefits, who could be harmed?**

**Answer:** 

**Stakeholder:**  a national environmental agency that compiles the greenhouse-gas inventory and plans peatland restoration (for example, rewetting drained peatlands), or a research consortium advising it.

**Their goals:**  know how much CH4 a peatland emits without months of chamber measurements at every site, and plan rewetting in a way that cuts CO2 and N2O from drained peat without creating large new CH4 sources.

**Concrete decision supported:** whether it is worth adding soil DNA (metagenome) analysis to standard peatland site surveys, or whether cheaper soil and water measurements (water table, moisture, chemistry) give equally good CH4 estimates. A secondary decision is which site conditions to monitor to flag likely high-CH4 sites.

**Who benefits:** the agency and taxpayers (lower survey costs, better inventory estimates), climate policy, and researchers who learn whether microbes carry extra predictive information.

**Who could be harmed:** landowners and farmers, if wrong estimates lead to restrictions, payments or rewetting decisions for their land; regions with few sites in the data (some land uses have only 2–4 sites), whose emissions may be estimated worse; and the climate, if CH4 after rewetting is under-estimated. Money is wasted if DNA analysis is adopted although it does not improve the estimates.

**Answer:**

**Objective:** Estimate the methane emission of a peatland that has never been measured from a single field visit, and find out whether paying for soil-microbe DNA analysis makes these estimates clearly better than soil and water measurements alone. A secondary objective is to connect site factors (wetness, chemistry, microbes) to the level of emissions.

**How success would be judged:**

* The estimates for sites that were not used to build the estimate must be clearly better than simply using the average emission of other peatlands.

* DNA analysis is "worth it" only if the improvement it brings is larger than the variation of the error between different groups of test sites; otherwise the money is better spent on measuring more sites.
* The factors identified as important should agree with known peatland processes (e.g. a high water table favouring CH4) and stay the same when reasonable analysis choices change.

**d. Now translate the objective to a data science problem.**

Is it classification, regression, clustering, description? What would be predicted or explained? At what point in time would the prediction or other outcome be used?

**Answer:**

**Type:** supervised regression (plus model interpretation). The unit is one station.
Predicted: the CH4 flux of a station on the log10(CH4 + 100) scale (CH4_log10_plus100), a continuous value. The constant 100 allows zero and negative fluxes (uptake down to about −65).

**Predictors:** (1) soil moisture only (baseline), (2) the 16 environmental variables, (3) the 16 variables plus methanogen_share_pct and methanotroph_share_pct, and optionally (4) all 210 genera (proportions or CLR). Comparing (2) and (3) answers the main question.

**Evaluation:** cross-validation by site (all stations of a site in the same fold), so every score is on unseen sites; R² and RMSE, and the spread of RMSE across folds. Feature importance (SHAP) explains which predictors drive the predictions.

**When it is used:** after a one-off field visit to a new or planned restoration site, once soil samples and water-table/moisture readings (and optionally a DNA sample) are available, and before the decision on monitoring or rewetting. The prediction replaces a long chamber-measurement campaign for screening; it describes conditions at the time of sampling, not a yearly total.

Site coordinates may point to private land, so we do not publish station-level maps or coordinates.

**Technical:** missing values must be imputed inside cross-validation to avoid leakage; results must run from start to finish.

**Ethics:** the model should support, not replace, field measurements in decisions about someone's land; regions with few sites (e.g. tropics and some land uses) may be represented poorly, and we will report this limitation.

```python
## TODO
# !pip install matplotlib
import matplotlib.pyplot as plt

df = pd.read_csv("MIEF-PEAT_CH4_dataset_v0.csv")
y = "CH4_log10_plus100"
d = df.dropna(subset=[y])

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].hist(d["CH4_ugC_m2_h"], bins=40)
ax[0].set(title="Raw CH4 flux", xlabel="µg C m⁻² h⁻¹")
ax[1].hist(d[y], bins=30)
ax[1].set(title="log10(CH4 + 100)", xlabel="target")
plt.tight_layout()
plt.show()

order = d.groupby("land_use")[y].median().sort_values().index
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.boxplot([d.loc[d.land_use == k, y] for k in order],
           tick_labels=[f"{k}\n(n={(d.land_use == k).sum()})" for k in order])
ax.set(ylabel="log10(CH4 + 100)", title="Target by land use")
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
s = ax[0].scatter(d.water_table_cm, d[y], c=d.soil_moisture_vwc, cmap="viridis", s=18)
ax[0].set(xlabel="water table (cm)", ylabel="target")
plt.colorbar(s, ax=ax[0], label="soil moisture")
ax[1].scatter(d.soil_moisture_vwc, d[y], s=18)
ax[1].axvline(1, color="r", ls="--")
ax[1].set(xlabel="soil moisture (vwc)")
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
ax[0].scatter(d.methanogen_share_pct, d[y], s=18)
ax[0].set(xlabel="methanogen share (%)", ylabel="target")
ax[1].scatter(d.methanotroph_share_pct, d[y], s=18)
ax[1].set(xlabel="methanotroph share (%)")
plt.tight_layout()
plt.show()

env = ["pH", "von_post_grade", "organic_matter_frac", "soil_moisture_vwc",
       "P_mg_kg", "K_mg_kg", "Ca_mg_kg", "Mg_mg_kg", "nitrate_mg_kg",
       "ammonium_mg_kg", "total_N_pct", "total_C_pct", "CN_ratio",
       "water_table_cm", "soil_temp_10cm_C", "soil_temp_20cm_C"]
cols = [y] + env + ["methanogen_share_pct", "methanotroph_share_pct"]
corr = d[cols].corr(method="spearman")
fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
ax.set_xticks(range(len(cols)))
ax.set_xticklabels(cols, rotation=90)
ax.set_yticks(range(len(cols)))
ax.set_yticklabels(cols)
plt.colorbar(im)
plt.tight_layout()
plt.show()
```

Does the target measure what the objective needs? Mostly yes, with one limit. Our objective is to estimate how much methane a peatland emits, and the target is the CH4 flux measured with closed chambers at the station, so it is a direct measurement of emission and not a proxy. The limits are these:
* Each station has one flux value (the dictionary calls it the "mean CH4 flux of the station"), so the target is a snapshot or short-term mean. It describes the emission at the time of sampling, not a yearly total, and CH4 changes a lot with season and weather. We can use the model for screening and comparing sites, but not for a yearly inventory number, unless it is validated against annual measurements.
* The target is very skewed: the median flux is about 253 µg C m⁻² h⁻¹, but the maximum is about 37,900, and 33 stations have negative values (CH4 uptake). The log10(CH4 + 100) transformation handles this, but the constant 100 is provisional, and we will check how much the results depend on it.

Is the sample representative? Only partly.
* The data cover 75 sites worldwide (latitude −55° to 68°), and 20 sites are in the tropics and 20 in the southern hemisphere, so the coverage is wide. Some regions and ecosystems are probably missing.
